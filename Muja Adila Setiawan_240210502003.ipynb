```python
# ============================================================
# 1. KOLEKSI DOKUMEN
# ============================================================
judul_skripsi = [
    "Implementasi Keamanan Jaringan pada MikroTik OS Menggunakan Metode Port Knocking",
    "Rancang Bangun Jemuran Otomatis untuk Smart Home Berbasis IoT",
    "Perbandingan Performa SVM dalam Analisis Sentimen Komentar Israel-Palestina Menggunakan LSA dan Word2vec",
    "Implementasi Machine Learning untuk Klasifikasi Sentimen dan Pemodelan Topik Komentar Pengguna pada Video Ulasan Produk Gadget Berbasis Website",
    "Pengembangan Sistem Kontrol Lampu dan Pengaman Pintu Rumah dengan Monitoring Kamera Berbasis ESP32 Terintegrasi Aplikasi Android",
    "Analisis Pemanfaatan Kecerdasan Buatan Generatif dalam Memahami Motivasi Belajar Mahasiswa Berbasis Self-Determination Theory SDT",
    "Implementasi Jaringan Syaraf Tiruan untuk Penentuan Kematangan dan Ketebalan Daging Buah Pepaya Berdasarkan Ciri Visual Warna Kulit",
    "Analisis dan Klasifikasi Kualitas Layanan Jaringan Wireless LAN dengan Algoritma Naive Bayes",
    "Analisis Efektivitas iptables dalam Melindungi Jaringan dari Serangan DDoS",
    "Klasifikasi Tanda Tangan Menggunakan Algoritma Convolutional Neural Network CNN"
]
# ============================================================
# GROUND TRUTH
# ============================================================
# Kunci jawaban relevansi yang ditentukan secara manual
ground_truth = {
    "AI": {3, 4, 6, 7, 10},
    "Jaringan": {1, 7, 8, 9},
    "IoT": {2, 5}
}
# ============================================================
# HASIL PENCARIAN SISTEM
# ============================================================
# 3 dokumen teratas yang dikembalikan sistem
hasil_pencarian = {

    "AI": [6, 4, 10],

    "Jaringan": [1, 8, 9],

    "IoT": [2, 5, 7]
}
print("Data berhasil dimuat!")
print(f"Jumlah judul: {len(judul_skripsi)}")
print(f"Kueri uji: {list(ground_truth.keys())}")
```

```python
# ============================================================
# PRECISION@3
# ============================================================
def precision_at_k(hasil, relevan, k):
    """
    Menghitung Precision@K
    Precision = jumlah dokumen relevan yang ditemukan
                / total dokumen yang dikembalikan
    """
    if k == 0:
        return 0.0
    # Menghitung dokumen relevan pada K hasil teratas
    ditemukan = sum(
        1 for d in hasil[:k]
        if d in relevan
    )
    # Precision = dokumen relevan / K
    return ditemukan / k

# ============================================================
# HITUNG PRECISION@3 UNTUK SEMUA KUERI
# ============================================================
k = 3
print("=" * 60)
print(f"PERHITUNGAN PRECISION@{k}")
print("=" * 60)

for kueri in hasil_pencarian.keys():
    hasil = hasil_pencarian[kueri]
    relevan = ground_truth[kueri]
    # Hitung Precision@3
    p = precision_at_k(
        hasil,
        relevan,
        k
    )

    print(f"\nKueri: '{kueri}'")
    print(f"  Hasil 3 teratas: {hasil}")
    print(f"  Dokumen relevan: {sorted(relevan)}")
    print(f"  Precision@{k}: {p:.3f}")

    print(
        f"  ({sum(1 for d in hasil[:k] if d in relevan)} "
        f"dari {k} dokumen yang dikembalikan adalah relevan)"
    )

print("\nPrecision@3 selesai dihitung!")
```

```python
# ============================================================
# RECALL
# ============================================================
def recall(hasil, relevan):
    """
    Menghitung Recall
    Recall = jumlah dokumen relevan yang ditemukan
             / total dokumen relevan di koleksi
    """
    ditemukan = sum(
        1 for d in hasil
        if d in relevan
    )
    return (
        ditemukan / len(relevan)
        if len(relevan) > 0
        else 0.0
    )
# ============================================================
# HITUNG RECALL UNTUK SEMUA KUERI
# ============================================================

print("=" * 60)
print("PERHITUNGAN RECALL")
print("=" * 60)

for kueri in hasil_pencarian.keys():
    hasil = hasil_pencarian[kueri]
    relevan = ground_truth[kueri]
    # Hitung Recall
    r = recall(
        hasil,
        relevan
    )
    ditemukan = sum(
        1 for d in hasil
        if d in relevan
    )
    total = len(relevan)
    print(f"\nKueri: '{kueri}'")
    print(f"  Hasil pencarian: {hasil}")
    print(
        f"  Dokumen relevan di koleksi: "
        f"{sorted(relevan)} (total {total})"
    )
    print(
        f"  Relevan yang ditemukan: "
        f"{ditemukan} dari {total}"
    )
    print(f"  Recall = {r:.3f}")

print("\nRecall selesai dihitung!")
```

```python
# ============================================================
# F1-SCORE
# ============================================================
def f1_score(precision, recall):
    """
    Menghitung F1-Score
    F1 = 2 x (Precision x Recall)
         / (Precision + Recall)
    """
    if precision + recall == 0:
        return 0.0
    return (
        2 * precision * recall
        / (precision + recall)
    )

# ============================================================
# HITUNG F1 UNTUK SEMUA KUERI
# ============================================================
print("=" * 60)
print("PERHITUNGAN F1")
print("=" * 60)
for kueri in hasil_pencarian.keys():
    hasil = hasil_pencarian[kueri]
    relevan = ground_truth[kueri]
    k = 3
    # Ambil Precision dan Recall
    p = precision_at_k(
        hasil,
        relevan,
        k
    )
    r = recall(
        hasil,
        relevan
    )

    # Hitung F1
    f1 = f1_score(
        p,
        r
    )

    print(f"\nKueri: '{kueri}'")
    print(f"  Precision@{k} = {p:.3f}")
    print(f"  Recall        = {r:.3f}")
    print(f"  F1-Score      = {f1:.3f}")

    print(
        f"  Rumus: 2 x ({p:.3f} x {r:.3f}) "
        f"/ ({p:.3f} + {r:.3f}) = {f1:.3f}"
    )

print("\nF1-Score selesai dihitung!")
```

```python
# ============================================================
# AVERAGE PRECISION (AP)
# ============================================================

def average_precision(hasil, relevan):

    """
    Menghitung Average Precision (AP)

    AP = rata-rata Precision pada setiap posisi
         ditemukannya dokumen relevan
    """

    hit = 0
    total = 0.0

    # Periksa setiap dokumen berdasarkan peringkat
    for i, d in enumerate(
        hasil,
        start=1
    ):

        if d in relevan:

            hit += 1

            precision_at_i = hit / i

            total += precision_at_i

    # AP = total precision / jumlah dokumen relevan
    return (
        total / len(relevan)
        if len(relevan) > 0
        else 0.0
    )


# ============================================================
# HITUNG AP UNTUK SEMUA KUERI
# ============================================================

print("=" * 60)
print("PERHITUNGAN AVERAGE PRECISION (AP)")
print("=" * 60)

for kueri in hasil_pencarian.keys():

    hasil = hasil_pencarian[kueri]
    relevan = ground_truth[kueri]

    print(f"\nKueri: '{kueri}'")
    print(f"  Hasil pencarian: {hasil}")
    print(
        f"  Dokumen relevan: {sorted(relevan)}"
    )

    print("\n  Detail per posisi:")
    print(
        "  Posisi | Dokumen | Relevan? | Hit | "
        "Precision di posisi ini"
    )
    print(
        "  -------|---------|----------|-----|----------------------"
    )

    hit = 0
    details = []
    for i, d in enumerate(
        hasil,
        start=1
    ):
        if d in relevan:
            hit += 1
            p_at_i = hit / i
            details.append(p_at_i)
            print(
                f"  {i:6} | {d:7} | Ya       | "
                f"{hit:3} | {p_at_i:.3f}"
            )

        else:

            print(
                f"  {i:6} | {d:7} | Tidak    | "
                f"{hit:3} | -"
            )

    # Hitung AP
    ap = average_precision(
        hasil,
        relevan
    )

    print(
        f"\n  Total Precision di setiap titik relevan: "
        f"{details}"
    )

    print(
        f"  AP = "
        f"({' + '.join(f'{d:.3f}' for d in details)}) "
        f"/ {len(relevan)} = {ap:.3f}"
    )

print("\nAverage Precision (AP) selesai dihitung!")
```

```python
# ============================================================
# MEAN AVERAGE PRECISION (MAP)
# ============================================================

def mean_average_precision(ap_values):

    """
    Menghitung Mean Average Precision (MAP)

    MAP = rata-rata AP dari semua kueri
    """

    return (
        sum(ap_values) / len(ap_values)
        if len(ap_values) > 0
        else 0.0
    )


# ============================================================
# HITUNG MAP
# ============================================================

print("=" * 60)
print("PERHITUNGAN MEAN AVERAGE PRECISION (MAP)")
print("=" * 60)

ap_values = {}

for kueri in hasil_pencarian.keys():

    hasil = hasil_pencarian[kueri]
    relevan = ground_truth[kueri]

    ap = average_precision(
        hasil,
        relevan
    )

    ap_values[kueri] = ap


print("\nNilai AP setiap kueri:")

for kueri, ap in ap_values.items():

    print(
        f"  {kueri}: {ap:.3f}"
    )


# Hitung MAP
map_score = mean_average_precision(
    list(ap_values.values())
)

print(
    f"\nMAP = "
    f"({' + '.join(f'{ap:.3f}' for ap in ap_values.values())}) "
    f"/ {len(ap_values)}"
)

print(
    f"MAP = {map_score:.3f}"
)

print(
    "\nMean Average Precision (MAP) selesai dihitung!"
)
```

```python
# ============================================
# GROUND TRUTH DAN HASIL PENCARIAN
# ============================================

data_query = {
    "cinta": {
        "relevan": {2, 3, 5, 9},
        "hasil": [2, 7, 3, 1, 5]
    },

    "rindu": {
        "relevan": {1, 4, 6, 8},
        "hasil": [1, 5, 4, 2, 7]
    },

    "bahagia": {
        "relevan": {2, 5, 7, 10},
        "hasil": [3, 2, 7, 9, 5]
    }
}


# ============================================
# FUNGSI PRECISION@K
# ============================================

def precision_at_k(hasil, relevan, k):
    return sum(
        1 for d in hasil[:k]
        if d in relevan
    ) / k


# ============================================
# FUNGSI RECALL
# ============================================

def recall(hasil, relevan):
    cocok = sum(
        1 for d in hasil
        if d in relevan
    )

    return cocok / len(relevan)


# ============================================
# FUNGSI F1
# ============================================

def f1(p, r):
    if p + r == 0:
        return 0.0

    return 2 * p * r / (p + r)


# ============================================
# FUNGSI AVERAGE PRECISION (AP)
# ============================================

def average_precision(hasil, relevan):
    hit = 0
    total = 0.0

    for i, d in enumerate(hasil, start=1):

        if d in relevan:
            hit += 1
            total += hit / i

    return total / len(relevan)


# ============================================
# MENGHITUNG HASIL 3 QUERY
# ============================================

nilai_ap = []

print("=" * 60)
print("HASIL EVALUASI MESIN PENCARI")
print("=" * 60)

for query, data in data_query.items():

    relevan = data["relevan"]
    hasil = data["hasil"]

    # Precision@5
    p = precision_at_k(hasil, relevan, 5)

    # Recall
    r = recall(hasil, relevan)

    # F1
    f = f1(p, r)

    # Average Precision
    ap = average_precision(hasil, relevan)

    # Simpan AP untuk MAP
    nilai_ap.append(ap)

    print("\nQuery :", query)
    print("Ground Truth    :", sorted(relevan))
    print("Hasil Pencarian :", hasil)
    print("Precision@5     :", round(p, 4))
    print("Recall          :", round(r, 4))
    print("F1              :", round(f, 4))
    print("AP              :", round(ap, 4))


# ============================================
# MENGHITUNG MAP
# ============================================

MAP = sum(nilai_ap) / len(nilai_ap)

print("\n" + "=" * 60)
print("HASIL AKHIR")
print("=" * 60)
print("MAP :", round(MAP, 4))
print("=" * 60)
```
